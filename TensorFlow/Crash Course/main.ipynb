```python
import tensorflow as tf
```

```python
mnist = tf.keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_train[0]
```

```python
X_train, X_test = X_train / 255, X_test / 255
X_train[0]
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input, Dropout

# Dense -> A layer where a neuron from prev is connected to every neuron
# Build Model
model = Sequential()
model.add(Input((28, 28))) # input size ours is 28*28 pixels
model.add(Flatten())
model.add(Dense(128, activation='relu')) # clip min to 0
model.add(Dropout(0.2)) # Regularization Layer, 20% chance a connection will not be there
model.add(Dense(10))
```

```python
nonsense_prediction = model(X_train[:1]).numpy()
nonsense_prediction
```

```python
tf.nn.softmax(nonsense_prediction).numpy().sum()
```

```python
from tensorflow.keras.losses import SparseCategoricalCrossentropy 

loss_function = SparseCategoricalCrossentropy(from_logits=True)
# -tf.math.log(1/10)
loss_function(y_train[:1], nonsense_prediction)
```

```python
# compile model and train
from tensorflow.keras.optimizers import Adam

optimizer = Adam(learning_rate=0.01)

model.compile(optimizer=optimizer, loss=loss_function, metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10)
```

```python
model.evaluate(X_test, y_test) # returns [loss, accuracy]
```

```python
tf.nn.softmax(model(X_test[:1])).numpy()
```

# Regression Task
### Where you predict not 'cat' or 'dog' but give me the price of the laptop/stock, etc.

```python
import pandas as pd

df = pd.read_csv('housing.csv')
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X, y = df.drop('MedHouseVal', axis=1), df['MedHouseVal']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_test_scaled.shape
```

```python
model = Sequential()
model.add(Input((8,)))
model.add(Dense(64, activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(1))
```

```python
model.compile(optimizer='adam', loss='mse', metrics=['mae', 'r2_score'])
```

```python
model.fit(X_train_scaled, y_train, epochs=10)
```

```python
model.evaluate(X_test_scaled, y_test)
```
